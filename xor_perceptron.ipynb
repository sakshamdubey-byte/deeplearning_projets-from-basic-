```python
# let's try to implement or and xor using perceptron(single neural network )
import numpy as np

# Perceptron Class Definition
class Perceptron:
    def __init__(self, input_size, lr=0.1, epochs=100):
        self.weights = np.zeros(input_size)
        self.bias = 0
        self.lr = lr
        self.epochs = epochs

    def step_function(self, x):
        return 1 if x >= 0 else 0

    def predict(self, x):
        linear_output = np.dot(x, self.weights) + self.bias
        return self.step_function(linear_output)

    def train(self, X, y):
        for _ in range(self.epochs):
            for inputs, label in zip(X, y):
                prediction = self.predict(inputs)
                error = label - prediction
                # Update weights and bias
                # error positive weights will increase
                # error negatvei eirghts will decreas to compensate
                self.weights += self.lr * error * inputs
                self.bias += self.lr * error

# tinnnny data set fr
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# ye to basically  y ki og value hai
y_or  = np.array([0, 1, 1, 1])
y_and = np.array([0, 0, 0, 1])
y_xor = np.array([0, 1, 1, 0])


# helping funcion
def evaluate_gate(gate_name, X, y):
    model = Perceptron(input_size=2)
    model.train(X, y)
    print(f"--- {gate_name} Gate ---")
    for inputs in X:
        print(f"Input: {inputs} -> Predicted: {model.predict(inputs)}")
    print()

# Run tests
evaluate_gate("OR", X, y_or)
evaluate_gate("AND", X, y_and)
# it will give wrong outputs  in xor bcz that is not linearly seperable
evaluate_gate("XOR", X, y_xor)
```
